#SQLITE

```python
import sqlite3
conn=sqlite3.connect("students.db")
```

```python
cursor=conn.cursor()
print("Database Created Successfully")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS student_table(
NAME TEXT,
ROLL_NO INTEGER,
MARKS INTEGER)

""")

print("Table is created Successfully !!")
```

```python
#write the data into table
cursor.execute("INSERT INTO student_table VALUES('Astik',1,85)")
cursor.execute("INSERT INTO student_table VALUES('Abhiyank',2,88)")
cursor.execute("INSERT INTO student_table VALUES('Aditya',3,98)")
cursor.execute("INSERT INTO student_table VALUES('Ayansh',4,80)")
cursor.execute("INSERT INTO student_table VALUES('Ayush',5,80)")
```

```python
conn.commit()
print('Recordes inserted successfully !!')
conn.close()
```

```python
#Retrive | read the data
conn=sqlite3.connect("students.db")
cursor=conn.cursor()
cursor.execute("SELECT * FROM student_table")
rows=cursor.fetchall()
for row in rows:
  print(row)
  conn.close()
```

Q.2  Write a file using ; as the delimiter and -999 for missing values.Read it so -999 becomes NaN.


```python
data_new = {
    "Name": ["Prakash", "Rahul", "Aman"],
        "Marks": [85, -999, 90],
         "Age": [20, 21, -999]
     }
tem_df=pd.DataFrame(data_new)
tem_df.to_csv("students.csv", sep=";", index=False, na_rep="-999")
tem_df = pd.read_csv("students.csv", sep=";", na_values="-999")
print(tem_df)
```
